```python
import os
from git import Repo

# 第一步：克隆仓库
repo_url = "https://github.com/facebookresearch/fine_grained_hateful_memes.git"
repo_dir = "fine_grained_hateful_memes"  # 临时目录用于克隆仓库

if not os.path.exists(repo_dir):
    Repo.clone_from(repo_url, repo_dir)

# 第二步：移动并重命名 'annotations' 文件夹
source_folder = os.path.join(repo_dir, "data", "annotations")
destination_folder = os.path.join(data_dir, "hateful_memes_finegrained")  # 直接放入已存在的 'data' 文件夹

if os.path.exists(source_folder):
    # 将文件夹重命名并移动到 'data' 文件夹
    os.rename(source_folder, destination_folder)
    print(f"The folder was successfully moved and renamed: {destination_folder}")
else:
    print("Not found in the repo 'annotations', File folder (renamed: hateful_memes_finegrained)")
```
