```python
import requests

repo_url = "https://github.com/facebookresearch/fine_grained_hateful_memes"

try:
    response = requests.get(repo_url)
    if response.status_code == 200:
        print("✅ 仓库是公开的，可以访问")
    elif response.status_code == 404:
        print("❌ 仓库不存在或已改名")
    else:
        print(f"⚠️ 访问受限，状态码: {response.status_code}")
except Exception as e:
    print(f"❌ 访问失败: {e}")
```

```python
import os
from git import Repo

# Step 1: Clone the repository
repo_url = "https://github.com/facebookresearch/fine_grained_hateful_memes"
repo_dir = "fine_grained_hateful_memes"  # Temporary directory to clone the repo

if not os.path.exists(repo_dir):
    Repo.clone_from(repo_url, repo_dir)

# Step 2: Move and rename the annotations folder
source_folder = os.path.join(repo_dir, "data", "annotations")
destination_folder = os.path.join("data", "hateful_memes_finegrained")  # Directly in the existing 'data' folder

if os.path.exists(source_folder):
    # Rename and move the folder to the existing 'data' folder
    os.rename(source_folder, destination_folder)
    print(f"Successfully moved and renamed the folder to: {destination_folder}")
else:
    # print("仓库中没有找到 'annotations' 文件夹（重命名为 hateful_memes_finegrained）。")
    "The 'annotations' folder (renamed: hateful_memes_finegrained) does not exist in the repository."
```
