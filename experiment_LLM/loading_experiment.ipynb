```python
import subprocess

def describe_image_with_minigpt4(image_path):
    """
    使用 MiniGPT-4 对图像进行描述（通过 subprocess 调用命令行 evaluate.py）
    """
    prompt = "Describe this image in detail."
    command = [
        "python", "evaluate.py",
        "--cfg-path", "eval_configs/minigpt4_eval.yaml",
        "--gpu-id", "0",
        "--image-path", image_path,
        "--text-input", prompt
    ]
    result = subprocess.run(command, capture_output=True, text=True)
    output = result.stdout.strip()
    
    # 提取最后一句为描述（可根据实际 output 格式调整）
    lines = output.split('\n')
    description = lines[-1] if lines else ""
    return description
```

```python
import openai
import requests
from PIL import Image
from io import BytesIO
from transformers import BlipProcessor, BlipForConditionalGeneration
import pytesseract  # 用于OCR（文字识别）

# 设置API密钥和URL
API_SECRET_KEY = "sk-zk21884c49fc398427914f99fc30171ecb068f998dc6f05b"
BASE_URL = "https://api.zhizengzeng.com/v1/chat/completions"

# 加载BLIP模型和处理器
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base")

# 加载图像并进行处理
def load_image(image_path, resize=True):
    image = Image.open(image_path)
    if resize:
        image = image.resize((512, 512))  # 调整图像大小以加速处理
    return image

# 图像文字识别函数（OCR）
def extract_text_from_image(image_path):
    image = load_image(image_path)
    text = pytesseract.image_to_string(image)  # 使用tesseract进行OCR
    return text.strip()  # 返回图像中的文字内容

# 使用BLIP生成图像描述
def describe_image_with_blip(image_path):
    image = load_image(image_path)
    
    # 图像处理
    inputs = processor(image, return_tensors="pt")
    
    # 生成图像描述
    out = model.generate(**inputs)
    description = processor.decode(out[0], skip_special_tokens=True)
    return description

# 生成分析内容的Prompt
def generate_analysis_prompt(image_path, description):
    # 提取图像中的文本
    text_from_image = extract_text_from_image(image_path)
    
    # 构造用于生成分析的Prompt，结合图像描述与图像中的文字
    prompt = (
        f"请分析以下图像内容，结合图像中的文字：\n"
        f"图像描述：{description}\n"
        f"图像中的文字：{text_from_image}\n"
        f"请简洁描述图像中的不良内容，涉及歧视、暴力、血腥、误导性等。"
    )
    return prompt

# 进一步分析描述
def analyze_caption_with_gpt4(description, image_path):
    prompt = generate_analysis_prompt(image_path, description)
    
    analysis_url = "https://api.zhizengzeng.com/v1/chat/completions"
    
    headers = {
        'Content-Type': 'application/json',
        'Authorization': f'Bearer {API_SECRET_KEY}',
    }
    
    data = {
        "model": "gpt-4.1",
        "messages": [
            {
                "role": "user",
                "content": prompt
            }
        ]
    }
    
    response = requests.post(analysis_url, headers=headers, json=data)
    
    if response.status_code == 200:
        result = response.json()
        analysis = result["choices"][0]["message"]["content"]
        return analysis
    else:
        return f"Error: {response.status_code}, {response.text}"

# 主函数示例：上传图像并分析描述
def main(image_path):
    # 使用BLIP生成图像描述
    description = describe_image_with_blip(image_path)
    print("生成的描述：", description)

    # 分析图像描述
    analysis = analyze_caption_with_gpt4(description, image_path)
    print("分析结果：", analysis)

# 示例图片路径，替换为您的图片路径
image_path = "/Users/suleynan_suir/Desktop/NLP_Package/Project/Hateful-Image-Project/data/img/01327.png"  # 替换为您的图片路径
main(image_path)
```

```python
import openai
import requests
import time
from PIL import Image
from io import BytesIO
from transformers import BlipProcessor, BlipForConditionalGeneration
import pytesseract  # 用于OCR（文字识别）

# 设置API密钥和URL
API_SECRET_KEY = "sk-zk21884c49fc398427914f99fc30171ecb068f998dc6f05b"
BASE_URL = "https://api.zhizengzeng.com/v1/chat/completions"

# 加载BLIP模型和处理器
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base")

# 加载图像并进行处理
def load_image(image_path, resize=False):
    image = Image.open(image_path)
    if resize:
        image = image.resize((512, 512))  # 调整图像大小以加速处理
    return image

# 图像文字识别函数（OCR）
def extract_text_from_image(image_path):
    image = load_image(image_path)
    text = pytesseract.image_to_string(image)  # 使用tesseract进行OCR
    return text.strip()  # 返回图像中的文字内容

# 使用BLIP生成图像描述
def describe_image_with_blip(image_path):
    image = load_image(image_path)
    
    # 图像处理
    inputs = processor(image, return_tensors="pt")
    
    # 生成图像描述
    out = model.generate(**inputs)
    description = processor.decode(out[0], skip_special_tokens=True)
    return description

# 生成分析内容的Prompt
def generate_analysis_prompt(image_path, description):
    # 提取图像中的文本
    text_from_image = extract_text_from_image(image_path)
    
    # 构造用于生成分析的Prompt，结合图像描述与图像中的文字
    prompt = (
        f"请分析以下图像内容，结合图像中的文字：\n"
        f"图像描述：{description}\n"
        f"图像中的文字：{text_from_image}\n"
        f"请详细分析图像中可能包含的不良要素，涉及：\n"
        f"1. 歧视性言论或行为\n"
        f"2. 暴力或血腥元素\n"
        f"3. 误导性信息或虚假内容\n"
        f"4. 种族、民族、性别或文化背景的歧视\n"
        f"5. 其他可能的不良或有害内容\n"
        f"分析应结合图片中的背景和上下文，指出潜在的恶意信息和不当语言的深层含义。"
    )
    return prompt

# 进一步分析描述
def analyze_caption_with_gpt4(description, image_path):
    prompt = generate_analysis_prompt(image_path, description)

    analysis_url = "https://api.zhizengzeng.com/v1/chat/completions"
    
    headers = {
        'Content-Type': 'application/json',
        'Authorization': f'Bearer {API_SECRET_KEY}',
    }
    
    data = {
        "model": "gpt-4.1",
        "messages": [
            {
                "role": "user",
                "content": prompt
            }
        ]
    }

    # 开始计时
    start_time = time.time()

    # 发起请求
    response = requests.post(analysis_url, headers=headers, json=data)

    # 结束计时
    end_time = time.time()

    # 计算推理时间
    inference_time = end_time - start_time

    # 计算token数和费用
    tokens_used = response.json().get('usage', {}).get('total_tokens', 0)
    cost = (tokens_used / 1000) * 0.03  # 假设费用为每1000个token 0.03美元

    # 输出推理时间、token和费用
    print(f"推理时间: {inference_time:.2f}秒")
    print(f"消耗Token数量: {tokens_used}")
    print(f"推理费用: ${cost:.4f}")

    if response.status_code == 200:
        result = response.json()
        analysis = result["choices"][0]["message"]["content"]
        return analysis
    else:
        return f"Error: {response.status_code}, {response.text}"

# 主函数示例：上传图像并分析描述
def main(image_path):
    # 使用BLIP生成图像描述
    description = describe_image_with_blip(image_path)
    print("生成的描述：", description)

    # 分析图像描述
    analysis = analyze_caption_with_gpt4(description, image_path)
    print("分析结果：", analysis)

# 示例图片路径，替换为您的图片路径
image_path = "/Users/suleynan_suir/Desktop/NLP_Package/Project/Hateful-Image-Project/data/img/01327.png"  # 替换为您的图片路径
main(image_path)
```
