```python
import openai
from PIL import Image
import requests
from io import BytesIO
import torch
from torchvision import transforms
# from CLIP import clip  # 安装 CLIP 库
import clip

# 设置API密钥
# openai.api_key = "your-api-key-here"

# openai.api_key = "XXX"


# 加载 CLIP 模型
device = "cuda" if torch.cuda.is_available() else "cpu"
model, preprocess = clip.load("ViT-B/32", device)

# 加载图像
# image_path = "/path/to/your/image.jpg"
image_path = "/Users/suleynan_suir/Desktop/NLP_Package/Project/Hateful-Image-Project/data/img/01268.png"

image = Image.open(image_path)

# 图像预处理
image_input = preprocess(image).unsqueeze(0).to(device)

# 提取图像特征
with torch.no_grad():
    image_features = model.encode_image(image_input)

# 使用 CLIP 模型生成文本
text_input = torch.cat([clip.tokenize(["What does the image describe?"])]).to(device)

# 获取文本和图像之间的相似度
with torch.no_grad():
    text_features = model.encode_text(text_input)

# 计算图像与文本之间的相似度
similarity = (image_features @ text_features.T).squeeze(0)

print(f"Image similarity: {similarity.item()}")

# 使用 GPT-4 创建图像描述
response = openai.Completion.create(
    model="gpt-3.5-turbo",
    prompt=f"Describe what this image is about based on the features provided. Similarity score: {similarity.item()}",
    max_tokens=100
)

# 输出生成的图像描述
print(response.choices[0].text.strip())
```
